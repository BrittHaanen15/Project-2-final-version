```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Plotting iters vs. size for Jacobi algorithm - Tridiagonal matrix

```python
#Read file that my C++ script produced - tridia
data = pd.read_csv("../Project2_outputs/Size_and_iterations_tridiagonal.txt", sep=' ', header=None)

N_t = data[0]
iters_t = data[1]

#Read file that my C++ script produced - dense
data = pd.read_csv("../Project2_outputs/Size_and_iterations_dense.txt", sep=' ', header=None)

N_d = data[0]
iters_d = data[1]

#Plot
plt.plot(N_t, iters_t, label='Tridiagonal')
plt.plot(N_d, iters_d, label='Dense')
# plt.yscale('log')
# plt.xscale('log')
plt.title('Iterations needed as function of size, tridiagonal matrix')
plt.ylabel('Iterations')
plt.xlabel('N (matrix size)')
plt.legend()
plt.savefig("Iters_vs_N.pdf")
```

```python
#Plot
plt.plot((N_t)**2, iters_t, label='Tridiagonal')
plt.plot((N_d)**2, iters_d, label='Dense')
plt.yscale('log')
plt.xscale('log')
plt.title('Iterations needed as function of size, tridiagonal matrix')
plt.ylabel('Iterations')
plt.xlabel(f'N$^2$')
plt.legend()
plt.savefig("Iters_vs_N_squared.pdf")
```

# Plotting eigenvectors buckling beam, n=10 case

```python
#Read file that my C++ script produced - n=10
data = pd.read_csv("../Project2_outputs/Discrete_solution_n10.txt", sep=r'\s+', header=None)

#Extract variables
x = data[0]
vec_1 = data[1]
vec_2 = data[2]
vec_3 = data[3]

plt.scatter(x, vec_1, c='blue', label='First allowed eigenstate')
plt.plot(x, vec_1, linestyle=':', c='blue', label='_First allowed eigenstate')
plt.scatter(x, vec_2, c='red', label='Second allowed eigenstate')
plt.plot(x, vec_2, linestyle=':', c='red', label='_First allowed eigenstate')
plt.scatter(x, vec_3, c='green', label='Third allowed eigenstate')
plt.plot(x, vec_3, linestyle=':', c='green', label='_First allowed eigenstate')
plt.legend(bbox_to_anchor=(1.0, 0.5), loc="center left")
plt.xlabel(r'$\hat{x}$')
plt.ylabel(r'$v(\hat{x})$')
plt.title('The first three eigenvectors for buckling beam, n=10')
plt.savefig("Buckling beam states n10.pdf", bbox_inches='tight')
```
